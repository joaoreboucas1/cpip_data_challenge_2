```python
import numpy as np
import getdist
from getdist import plots
```

```python
CHAINS_PATH = "../../chains/"
CHAINS_STEM = "BPCA_VAL_"
indices = range(1, 4)

chains = {}

for i in indices:
    chain = getdist.loadMCSamples(CHAINS_PATH + CHAINS_STEM + str(i), settings={"ignore_rows": 0.3})
    chain.addDerived(chain["sigma8"]*np.sqrt(chain["omegam"]/0.3), name="S8", label="S_8")
    chains[i] = chain
```

```python
p = plots.get_subplot_plotter()
p.triangle_plot(
    [chains[i] for i in indices],
    ["omegam", "S8", "roman_BARYON_Q1", "roman_BARYON_Q2"],
    filled=[False, True, False],
    contour_colors=["orange", "lightblue", "green"],
    legend_labels=["BAHAMAS-T78", "DMO", "OWLS-AGN-T85"],
    markers={"omegam": 0.3, "S8": 0.821}
)
p.fig.suptitle("Model: PCA Marginalization", y=1.05)
p.export("plots/with_pca_marginalization.pdf")
```
