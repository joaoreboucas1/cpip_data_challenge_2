```python
import numpy as np
import cobaya
from cobaya.model import get_model
import cosmolike_roman_real_interface as cosmolike
import euclidemu2
CWD = "./projects/cpip_data_challenge_2/results/chi_by_eye/"
```

```python
problematic_chain_indices = [2, 4, 6, 7, 10, 11]
```

```python
model = get_model(f"{CWD}/TEMPLATE.yaml")
sampled_params_names = model.parameterization.sampled_params()
```

```python
# Random Sampling
THRESHOLD = 20_000
for _ in range(300):
    random_point = model.prior.sample()[0]
    random_point_dict = dict(zip(sampled_params_names, random_point))
    statistics = model.loglikes(random_point_dict, as_dict=True)
    chi2 = -2*statistics[0]["roman_real.cosmic_shear"]
    if chi2 < THRESHOLD:
        print(f"Good point (chi2 = {chi2:.3f}): {random_point_dict}")
    
```

```python
# Individual parameter adjustment
for omegam in np.linspace(0.2, 0.6, 10):
    random_point_dict["omegam"] = omegam
    statistics = model.loglikes(random_point_dict, as_dict=True)
    chi2 = -2*statistics[0]["roman_real.cosmic_shear"]
    print(f"\\Omega_m = {omegam:.3f} => chi2 = {chi2:.3f}")
```

```python
# Individual parameter adjustment
random_point_dict["omegam"] = 0.422
for As_1e9 in np.linspace(0.5, 5, 10):
    random_point_dict["As_1e9"] = As_1e9
    statistics = model.loglikes(random_point_dict, as_dict=True)
    chi2 = -2*statistics[0]["roman_real.cosmic_shear"]
    print(f"As_1e9 = {As_1e9:.3f} => chi2 = {chi2:.3f}")
```
